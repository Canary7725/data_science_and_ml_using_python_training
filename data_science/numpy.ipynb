```python
matrix_a=np.array([[1,2,3],
                   [2,4,6],
                   [3,6,9]])
            
matrix_b=np.array([[10,20,30],
                   [12,14,16],
                   [23,26,29]])

print(matrix_a + matrix_b)
print(matrix_a - matrix_b)
print(matrix_a / matrix_b)
print(matrix_a * matrix_b)
print(matrix_a @ matrix_b) #Matrix multiplication is done by @ but direct index-to-index multiplication is done by *
```

```python
#variance ==> How much the data scatters from the center (best-fit)

data1=np.array([48,49,50,51,52])
data2=np.array([10,30,50,70,90])


print(data1.mean())
print(data2.mean())

print(np.std(data1))
print(np.std(data2))


#Skewness ==> How much a data leans towards one side or another
#[1,2,3,4,5] ==> Balanced
#[1,91,92,93,94,95] ==> Negatively skewed
#[1,2,3,4,100] ==> Positively skewed
```

```python
# Correlation --> How much a particular affect another variable.

#var1 = Study hours
#var2 = Result

# -1 <--> 1
#-1 ==> Negative correlation
#0 ==> No correlation at all
#1 ==> Postive Correlation 

var_study_hours=np.array([1,2,3,4,5,6])
var_no_of_absetees=np.array([8,6,5,2,1,0])
var_shoe_size=np.array([6,7,5,4,8,5])
var_results=np.array([50,55,62,78,85,96])

correlation_matrix=np.corrcoef(var_no_of_absetees,var_results)
correlation_coefficient=correlation_matrix[0][1]

print(correlation_coefficient)
```
