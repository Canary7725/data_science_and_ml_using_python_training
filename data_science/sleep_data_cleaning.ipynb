```python
import pandas as pd
```

```python
df=pd.read_csv("data/sleep_dataset.csv")
```

```python
df.shape
```

```python
df.info()
```

```python
columns_to_be_renamed=df.columns.to_list()
column_name_mapper={}

for column in columns_to_be_renamed:
    column_name_mapper[column]=column.lower().replace(" ","_")
df=df.rename(columns=column_name_mapper)

df
```

```python
df['sleep_disorder']=df['sleep_disorder'].fillna("No Sleep Order Recorded")
```

```python
df.info()
```

```python
#age_group || 0-12-->Kid 13-19-->Teen 20-32 -->Young Adult 32-45 --> Adult 45-60-->Elder 60+ Old
```

```python
df.loc[df['age']<=12,'age_group']='Kid'
df.loc[(df['age']<=19) & (df['age']>12) ,'age_group']='Teen'
df.loc[(df['age']<=32) & (df['age']>19),'age_group']='Young Adult'
df.loc[(df['age']<=50) & (df['age']>32),'age_group']='Adult'
df.loc[df['age']>50,'age_group']='Elder'

df.info()


```

```python
df.loc[df['bmi_category'].isin(['Normal Weight']),'bmi_category']='Normal'
```

```python
#gender, occupation, bmi_category, sleep_disorder, age_group (category)
df[['gender','occupation','bmi_category','sleep_disorder','age_group']]=df[['gender','occupation','bmi_category','sleep_disorder','age_group']].astype("category")

df.dtypes
```

```python
# 130/85 --> Systolic=130 && Diastolic=85

# Optimal Blood Pressure --120/80 but 110-120(Normal) || 70-80(Normal)
```

```python
def get_systolic(value): #value = '130/85'
    split_value=value.split("/") # -->['130','85']
    return split_value[0]

def get_disatolic(value):
    split_value=value.split("/")
    return split_value[1]

df['systolic']=df['blood_pressure'].apply(get_systolic)
df['diastolic']=df['blood_pressure'].apply(get_disatolic)

df[['systolic','diastolic']]=df[['systolic','diastolic']].astype("int")

df.dtypes
```

```python
# 110-120(Normal) || 70-80(Normal)
df.loc[df['systolic'] > 120 ,'systolic_indicator']='High'
df.loc[df['systolic']<110,'systolic_indicator']='Low'

df.loc[(df['systolic']<=120) & (df['systolic']>=110),'systolic_indicator']='Normal'

df.loc[df['diastolic'] > 80 ,'diastolic_indicator']='High'
df.loc[df['diastolic']<70,'diastolic_indicator']='Low'

df.loc[(df['diastolic']>=70) & (df['diastolic']<=80),'diastolic_indicator']='Normal'

```

```python
#systolic !=Normal then final_indicator=systolic_indicator

#systolic ==Normal then final_indicator=diastolic_indicator

df.loc[df['systolic_indicator']!='Normal','final_indicator']=df['systolic_indicator']
df.loc[df['systolic_indicator']=='Normal','final_indicator']=df['diastolic_indicator']


df['final_indicator'].value_counts()
```
