# RNN:
Recurrent Neural Networks

Contains a layer that carrier past information. Mostly used for Speech and Text pattern recognition in Neural Networks.
Uses LSTM (Long Short-Term Memory) module for recording the memories.

```python
import pandas as pd
import re # Regular Expression
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
df=pd.read_csv('data/IMDB Dataset.csv')

df.head()
```

```python
def clean_text(text):
    text=text.lower()
    text= re.sub(r"[^a-zA-Z\s]","",text)
    return text
```

```python
df['review']=df['review'].apply(clean_text)
df.head()
```

```python
df['sentiment']=df['sentiment'].map({
    "positive":1,
    "negative":0
})
```

```python
df.head()
```

```python
tokenizer=Tokenizer(
    num_words=5000,
    oov_token='<OOV>'
)

# 0 --> padding
# 1 --> OOV tokens (Out of Vocaublary)
# 2 - 4999 --> distinct words
tokenizer.fit_on_texts(df['review'])
```

```python
sequences=tokenizer.texts_to_sequences(df['review'])
```

```python
padded=pad_sequences(
    sequences,
    maxlen=200,
    padding='post'
)

```

```python
padded
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(padded,df['sentiment'])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, LSTM, Dropout
```

```python
model=Sequential([
    Embedding(5000,64,input_length=200),

    LSTM(64,return_sequences=False),

    Dropout(0.5),
    Dense(32,activation='relu'), #Hidden Layer
    Dense(1,activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history=model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_data=(x_test,y_test)
)
```

```python
text=df.loc[49997,'review']
# predict(text)
```

```python
seq=tokenizer.texts_to_sequences([text])
padded_seq=pad_sequences(seq, maxlen=200,padding='post')
padded_seq
```

```python
prediction=model.predict(padded_seq)  

if(prediction[0][0]<0.5 and prediction[0][0]>=0):
    print("Negative Review")
elif(prediction[0][0]>=0.5 and prediction[0][0]<=1):
    print("Postive Review")


#0-0.3 --> negative
#0.3-0.5 --> slightly negative
#0.5-0.7-->slightly postive
#0.7-1-->postive
```
