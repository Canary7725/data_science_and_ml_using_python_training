To build a good machine learning model:

1. Bias (Underfitting)
Error from wrong assumptions in the model.
Model is too simple -> Fails to capture patterns which isnot in the training data.


2. Variance (Overfitting)
Error from too much sensitivity in the model.
Model is too complex -> Captures unnecessary patterns which is in the training data.

Build a model with low bias + low variance.

Goal: Good Generalization

How to detect bias vs variance?

High Bias -> Underfitting
1. Training accuracy is very low
2. Testing accuracy is very low
3. Gap between training and testing accuracy is very small


High Variance -> Overfitting
1. Training accuracy is very high
2. Testing accuracy is very low
3. Gap between testing and training accuracy is also high

How to fix Bias?
1. Increasing model complexity
2. Train longer
3. Add more features (feature engineering)
4. Decrease regularization
5. Use non linear models
6. Use polynomial features


How to fix Variance?
1. Decrease model complexity
2. Get more data
3. Apply data augmentation
4. Use Random Forest or Neural Network
5. Train for smaller number of epochs.
