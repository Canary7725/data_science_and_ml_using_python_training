```python
# What is Decision Tree?

# Decision Tree is a supervies ML algorithm that splits the data into branches based on feature values to make a decision.

# - Works for classification and regression.
# - Non-linear, interpretable, handles both numerical and categorical variables.

# How does a decision tree work?

# 1. Choose the best value to split the data.
# 2. Split data into child nodes based on that feature.
# 3. Repeat until:
#     1. Pure Leaf (same class).
#     2. Max depth reached.
#     3. Minimum samples reached.

# Splitting Criteria:

# For classification:
# 1. Gini impurity
# 2. Entropy (Information Gain)

# Advantages:
# 1. Easy to visualize
# 2. No scaling needed
# 3. Captures non-linear patterns

# Disadvantages:
# 1. Easily overfit
# 2. Unable to handle small data changes (high variance)
```

```python
import pandas as pd

df=pd.read_csv('data/student_performance_clean.csv')
df.drop(columns=['Unnamed: 0.1','Unnamed: 0'],inplace=True)

df
```

```python
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix,accuracy_score

import matplotlib.pyplot as plt


X=df[['study_hours','attendance']]
Y=df['pass_status']

x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2)

model=DecisionTreeClassifier(max_depth=3)

model.fit(x_train,y_train)

plt.figure(figsize=(20,12))

plot_tree(
    model,
    feature_names=['Study Hours','Attendance'],
    class_names=['Pass','Fail']
)

plt.show()

```

```python
new_value=pd.DataFrame(
    {
        "study_hours":[6],
        "attendance":[89]
    }
    )
predicted=model.predict(new_value)
print(predicted)

```

```python
y_prediciton=model.predict(x_test)

print(confusion_matrix(y_prediciton,y_test))

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_prediciton)
```
